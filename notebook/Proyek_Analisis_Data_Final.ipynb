## 1. Menentukan Pertanyaan Bisnis

Pertanyaan bisnis disusun menggunakan pendekatan **SMART** agar analisis memiliki fokus, ukuran yang jelas, relevansi bisnis, dan batas waktu.


**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

#### Pemeriksaan nilai kategorikal dan konsistensi data

Kode berikut memeriksa apakah kode kategori berada dalam rentang yang diperbolehkan, apakah nilai variabel ternormalisasi berada pada rentang 0–1, apakah jumlah penyewaan sama dengan penjumlahan pengguna kasual dan terdaftar, serta apakah informasi tahun, bulan, dan hari sesuai dengan tanggal.


```python
valid_category_values = {
    "season": [1, 2, 3, 4],
    "yr": [0, 1],
    "mnth": list(range(1, 13)),
    "holiday": [0, 1],
    "weekday": list(range(0, 7)),
    "workingday": [0, 1],
    "weathersit": [1, 2, 3, 4],
    "hr": list(range(0, 24))
}

def assess_data_quality(df, dataset_name):
    data = df.copy()
    data["dteday"] = pd.to_datetime(data["dteday"], errors="coerce")

    print(f"\nPemeriksaan kualitas {dataset_name}")
    print("-" * 45)

    print("Nilai kategori di luar rentang yang diharapkan:")
    for col, allowed in valid_category_values.items():
        if col in data.columns:
            invalid_count = (~data[col].isin(allowed)).sum()
            print(f"{col}: {invalid_count}")

    print("\nNilai di luar rentang 0–1:")
    for col in ["temp", "atemp", "hum", "windspeed"]:
        invalid_count = (~data[col].between(0, 1)).sum()
        print(f"{col}: {invalid_count}")

    print("\nNilai negatif pada jumlah penyewaan:")
    for col in ["casual", "registered", "cnt"]:
        print(f"{col}: {(data[col] < 0).sum()}")

    print("\nKetidaksesuaian cnt dengan casual + registered:")
    print((data["cnt"] != data["casual"] + data["registered"]).sum())

    print("\nKetidaksesuaian tahun, bulan, dan weekday dengan tanggal:")
    year_mismatch = (data["yr"] != data["dteday"].dt.year - 2011).sum()
    month_mismatch = (data["mnth"] != data["dteday"].dt.month).sum()
    weekday_from_date = (data["dteday"].dt.dayofweek + 1) % 7
    weekday_mismatch = (data["weekday"] != weekday_from_date).sum()

    print("Tahun:", year_mismatch)
    print("Bulan:", month_mismatch)
    print("Weekday:", weekday_mismatch)

    expected_workingday = (
        (~data["weekday"].isin([0, 6])) & (data["holiday"] == 0)
    ).astype(int)
    print(
        "Ketidaksesuaian workingday:",
        (data["workingday"] != expected_workingday).sum()
    )

assess_data_quality(day_df, "day.csv")
assess_data_quality(hour_df, "hour.csv")
```

#### Pemeriksaan kandidat outlier menggunakan IQR

Metode Interquartile Range (IQR) digunakan untuk menandai nilai yang berada di luar batas statistik. Hasil ini merupakan **kandidat outlier**, bukan bukti bahwa data salah. Nilai penyewaan yang tinggi dapat mencerminkan lonjakan permintaan aktual sehingga perlu dinilai berdasarkan konteks bisnis dan rentang nilai yang masuk akal.


```python
def detect_iqr_outliers(df, dataset_name):
    numeric_cols = [
        col for col in [
            "temp", "atemp", "hum", "windspeed",
            "casual", "registered", "cnt"
        ] if col in df.columns
    ]

    results = []
    for col in numeric_cols:
        q1 = df[col].quantile(0.25)
        q3 = df[col].quantile(0.75)
        iqr = q3 - q1
        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr
        outlier_count = (
            (df[col] < lower_bound) | (df[col] > upper_bound)
        ).sum()

        results.append({
            "dataset": dataset_name,
            "kolom": col,
            "batas_bawah": lower_bound,
            "batas_atas": upper_bound,
            "jumlah_kandidat_outlier": int(outlier_count)
        })

    return pd.DataFrame(results)

outlier_day = detect_iqr_outliers(day_df, "day.csv")
outlier_hour = detect_iqr_outliers(hour_df, "hour.csv")

outlier_summary = pd.concat(
    [outlier_day, outlier_hour],
    ignore_index=True
)

display(outlier_summary)
```

### 3.3 Cleaning Data

Cleaning dilakukan secara konsisten pada kedua dataset. Tindakan yang diterapkan:

1. Menghapus baris duplikat.
2. Mengonversi tanggal ke tipe datetime dan kolom numerik ke tipe numerik.
3. Menangani nilai kosong atau nilai yang gagal dikonversi.
4. Memvalidasi rentang kode kategori.
5. Memvalidasi rentang variabel ternormalisasi dan jumlah penyewaan.
6. Memeriksa konsistensi `cnt = casual + registered`.
7. Memeriksa konsistensi tahun, bulan, weekday, dan workingday terhadap tanggal.
8. Melakukan mapping kategori pada tahap cleaning.
9. Menghapus kolom indeks `instant` dan mengganti nama kolom agar lebih mudah dipahami.

Nilai kandidat outlier IQR tetap dipertahankan apabila masih berada dalam rentang domain yang valid dan tidak menunjukkan kesalahan pencatatan.


```python
# Mapping kategori dilakukan pada tahap Cleaning Data
year_mapping = {0: 2011, 1: 2012}

season_mapping = {
    1: "Spring",
    2: "Summer",
    3: "Fall",
    4: "Winter"
}

month_mapping = {
    i: pd.Timestamp(2000, i, 1).strftime("%B")
    for i in range(1, 13)
}

weekday_mapping = {
    0: "Sunday",
    1: "Monday",
    2: "Tuesday",
    3: "Wednesday",
    4: "Thursday",
    5: "Friday",
    6: "Saturday"
}

weather_mapping = {
    1: "Clear / Few Clouds",
    2: "Mist / Cloudy",
    3: "Light Snow / Rain",
    4: "Heavy Rain / Snow / Fog"
}

workingday_mapping = {
    0: "Non-Working Day",
    1: "Working Day"
}

holiday_mapping = {
    0: "Not Holiday",
    1: "Holiday"
}
```

```python
def clean_bike_data(raw_df, dataset_name):
    df = raw_df.copy()

    rows_before = len(df)
    missing_values_before = int(df.isna().sum().sum())
    duplicates_removed = int(df.duplicated().sum())

    # 1. Menghapus baris duplikat
    df = df.drop_duplicates().copy()

    # 2. Mengonversi tipe data
    df["dteday"] = pd.to_datetime(df["dteday"], errors="coerce")

    numeric_columns = [
        col for col in df.columns if col != "dteday"
    ]
    for col in numeric_columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    # 3. Memastikan kolom wajib tidak kosong
    required_columns = [
        "season", "yr", "mnth", "holiday", "weekday",
        "workingday", "weathersit", "temp", "atemp",
        "hum", "windspeed", "casual", "registered", "cnt"
    ]
    if "hr" in df.columns:
        required_columns.append("hr")

    valid = (
        df["dteday"].notna()
        & df[required_columns].notna().all(axis=1)
    )

    # 4. Memvalidasi kode kategori
    for col, allowed_values in valid_category_values.items():
        if col in df.columns:
            valid &= df[col].isin(allowed_values)

    # 5. Memvalidasi variabel yang dinormalisasi
    for col in ["temp", "atemp", "hum", "windspeed"]:
        valid &= df[col].between(0, 1)

    # 6. Jumlah penyewaan tidak boleh negatif
    for col in ["casual", "registered", "cnt"]:
        valid &= df[col].ge(0)

    # 7. Memastikan total sesuai komponen penyewaan
    valid &= df["cnt"].eq(df["casual"] + df["registered"])

    # 8. Memastikan tahun, bulan, dan weekday sesuai tanggal
    valid &= df["yr"].eq(df["dteday"].dt.year - 2011)
    valid &= df["mnth"].eq(df["dteday"].dt.month)
    valid &= df["weekday"].eq(
        (df["dteday"].dt.dayofweek + 1) % 7
    )

    expected_workingday = (
        (~df["weekday"].isin([0, 6])) & df["holiday"].eq(0)
    ).astype(int)
    valid &= df["workingday"].eq(expected_workingday)

    # Menghapus baris yang tidak memenuhi validasi
    invalid_rows_removed = int((~valid).sum())
    df = df.loc[valid].copy()

    # 9. Mapping kategori sebelum EDA
    df["year"] = df["yr"].map(year_mapping).astype(int)
    df["season_name"] = df["season"].map(season_mapping)
    df["month_number"] = df["mnth"].astype(int)
    df["month_name"] = df["month_number"].map(month_mapping)
    df["weekday_name"] = df["weekday"].map(weekday_mapping)
    df["holiday_status"] = df["holiday"].map(holiday_mapping)
    df["day_type"] = df["workingday"].map(workingday_mapping)
    df["weather_condition"] = df["weathersit"].map(weather_mapping)

    # 10. Mengganti nama kolom agar lebih mudah dipahami
    df = df.rename(columns={
        "dteday": "date",
        "hr": "hour",
        "temp": "temperature",
        "atemp": "feels_like_temperature",
        "hum": "humidity",
        "windspeed": "wind_speed",
        "casual": "casual_rentals",
        "registered": "registered_rentals",
        "cnt": "total_rentals"
    })

    # 11. Menghapus kolom indeks dan kode kategori yang sudah dimapping
    df = df.drop(
        columns=[
            "instant", "season", "yr", "mnth", "holiday",
            "weekday", "workingday", "weathersit"
        ],
        errors="ignore"
    )

    # Menata urutan kolom
    ordered_columns = [
        "date", "year", "month_number", "month_name",
        "season_name", "weekday_name", "holiday_status",
        "day_type", "weather_condition"
    ]

    if "hour" in df.columns:
        ordered_columns.append("hour")

    ordered_columns += [
        "temperature", "feels_like_temperature", "humidity",
        "wind_speed", "casual_rentals", "registered_rentals",
        "total_rentals"
    ]

    df = df[ordered_columns].sort_values(
        ["date"] + (["hour"] if "hour" in df.columns else [])
    ).reset_index(drop=True)

    cleaning_report = {
        "dataset": dataset_name,
        "rows_before": rows_before,
        "rows_after": len(df),
        "duplicates_removed": duplicates_removed,
        "invalid_rows_removed": invalid_rows_removed,
        "missing_values_before": missing_values_before,
        "missing_values_after": int(df.isna().sum().sum())
    }

    return df, cleaning_report
```

```python
# Menerapkan cleaning pada kedua dataset
day_clean, report_day = clean_bike_data(day_df, "day.csv")
hour_clean, report_hour = clean_bike_data(hour_df, "hour.csv")

cleaning_summary = pd.DataFrame([report_day, report_hour])
display(cleaning_summary)

print("Lima data pertama day_clean:")
display(day_clean.head())

print("Lima data pertama hour_clean:")
display(hour_clean.head())
```

## 4. Exploratory Data Analysis (EDA)

EDA dilakukan menggunakan `day_clean` dan `hour_clean`, yaitu dataset yang telah melalui proses cleaning dan mapping kategori. Analisis disusun untuk menjawab empat pertanyaan bisnis.


### 4.1 Pertanyaan 1: Pola Penyewaan Berdasarkan Bulan

Analisis ini membandingkan total dan rata-rata penyewaan setiap bulan pada tahun 2011 dan 2012. Total menunjukkan volume penyewaan, sedangkan rata-rata harian membantu membandingkan bulan dengan jumlah hari yang berbeda.


```python
monthly_summary = (
    day_clean
    .groupby(["year", "month_number", "month_name"], as_index=False)
    .agg(
        total_rentals=("total_rentals", "sum"),
        average_daily_rentals=("total_rentals", "mean")
    )
    .sort_values(["year", "month_number"])
)

year_summary = (
    day_clean.groupby("year", as_index=False)
    .agg(
        total_rentals=("total_rentals", "sum"),
        average_daily_rentals=("total_rentals", "mean")
    )
)

display(monthly_summary)
display(year_summary)

highest_month_2012 = (
    monthly_summary[monthly_summary["year"] == 2012]
    .loc[lambda x: x["total_rentals"].idxmax()]
)

lowest_month_2012 = (
    monthly_summary[monthly_summary["year"] == 2012]
    .loc[lambda x: x["total_rentals"].idxmin()]
)

print("Bulan tertinggi 2012:", highest_month_2012["month_name"],
      f"({highest_month_2012['total_rentals']:,} penyewaan)")
print("Bulan terendah 2012:", lowest_month_2012["month_name"],
      f"({lowest_month_2012['total_rentals']:,} penyewaan)")
```

### 4.2 Pertanyaan 2: Hubungan Cuaca, Suhu, dan Kelembapan

Analisis ini membandingkan rata-rata penyewaan berdasarkan kategori cuaca. Suhu dan kelembapan dikelompokkan menggunakan binning agar pola rata-rata penyewaan pada rentang nilai yang berbeda dapat diamati.


```python
weather_summary = (
    day_clean.groupby("weather_condition", as_index=False)
    .agg(
        total_rentals=("total_rentals", "sum"),
        average_rentals=("total_rentals", "mean"),
        number_of_days=("total_rentals", "size")
    )
    .sort_values("average_rentals", ascending=False)
)

day_environment = day_clean.copy()
day_environment["temperature_group"] = pd.cut(
    day_environment["temperature"],
    bins=4,
    include_lowest=True
)
day_environment["humidity_group"] = pd.cut(
    day_environment["humidity"],
    bins=4,
    include_lowest=True
)

temperature_summary = (
    day_environment.groupby("temperature_group", observed=True, as_index=False)
    .agg(average_rentals=("total_rentals", "mean"),
         number_of_days=("total_rentals", "size"))
)

humidity_summary = (
    day_environment.groupby("humidity_group", observed=True, as_index=False)
    .agg(average_rentals=("total_rentals", "mean"),
         number_of_days=("total_rentals", "size"))
)

display(weather_summary)
display(temperature_summary)
display(humidity_summary)

environment_correlation = day_clean[
    ["temperature", "humidity", "total_rentals"]
].corr()

print("Korelasi Pearson:")
display(environment_correlation)
```

### 4.3 Pertanyaan 3: Pola Penyewaan Berdasarkan Jam dan Jenis Hari

Analisis ini membandingkan rata-rata penyewaan pada setiap jam antara hari kerja dan hari nonkerja. Perbandingan ini membantu mengidentifikasi waktu dengan kebutuhan armada yang berbeda.


```python
hourly_summary = (
    hour_clean.groupby(["day_type", "hour"], as_index=False)
    .agg(
        average_rentals=("total_rentals", "mean"),
        total_rentals=("total_rentals", "sum")
    )
)

hourly_pivot = hourly_summary.pivot(
    index="hour",
    columns="day_type",
    values="average_rentals"
)

display(hourly_pivot.round(2))

peak_hours = (
    hourly_summary.loc[
        hourly_summary.groupby("day_type")["average_rentals"].idxmax()
    ]
    .sort_values("day_type")
)

lowest_hours = (
    hourly_summary.loc[
        hourly_summary.groupby("day_type")["average_rentals"].idxmin()
    ]
    .sort_values("day_type")
)

print("Jam dengan rata-rata penyewaan tertinggi:")
display(peak_hours)

print("Jam dengan rata-rata penyewaan terendah:")
display(lowest_hours)
```

### 4.4 Pertanyaan 4: Pola Penyewaan Berdasarkan Musim

Analisis ini membandingkan total dan rata-rata penyewaan pada setiap musim. Data juga dikelompokkan berdasarkan tahun untuk melihat perubahan pola antartahun.


```python
season_order = ["Spring", "Summer", "Fall", "Winter"]

season_summary = (
    day_clean.groupby("season_name", as_index=False)
    .agg(
        total_rentals=("total_rentals", "sum"),
        average_daily_rentals=("total_rentals", "mean"),
        number_of_days=("total_rentals", "size")
    )
)

season_summary["season_name"] = pd.Categorical(
    season_summary["season_name"],
    categories=season_order,
    ordered=True
)
season_summary = season_summary.sort_values("season_name")

season_year_summary = (
    day_clean.groupby(["year", "season_name"], as_index=False)
    .agg(
        total_rentals=("total_rentals", "sum"),
        average_daily_rentals=("total_rentals", "mean")
    )
)

season_year_summary["season_name"] = pd.Categorical(
    season_year_summary["season_name"],
    categories=season_order,
    ordered=True
)
season_year_summary = season_year_summary.sort_values(
    ["year", "season_name"]
)

display(season_summary)
display(season_year_summary)
```

### 4.5 Analisis Lanjutan (Opsional): Binning Tingkat Permintaan per Jam

Sebagai analisis tambahan tanpa machine learning, rata-rata penyewaan setiap jam dikelompokkan menjadi tiga tingkat permintaan menggunakan teknik binning. Pengelompokan ini membantu mengidentifikasi jam yang perlu mendapat prioritas kesiapan armada.


```python
hourly_overall = (
    hour_clean.groupby("hour", as_index=False)
    .agg(average_rentals=("total_rentals", "mean"))
)

lower_quantile, upper_quantile = hourly_overall[
    "average_rentals"
].quantile([1/3, 2/3])

hourly_overall["demand_level"] = pd.cut(
    hourly_overall["average_rentals"],
    bins=[-np.inf, lower_quantile, upper_quantile, np.inf],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)

display(hourly_overall.sort_values("hour"))

demand_level_summary = (
    hourly_overall.groupby("demand_level", observed=True, as_index=False)
    .agg(
        number_of_hours=("hour", "count"),
        average_rentals=("average_rentals", "mean")
    )
)

display(demand_level_summary)
```

## 6. Conclusion & Recommendation

### 6.1 Conclusion

**1. Pola Penyewaan Berdasarkan Bulan**

Total penyewaan meningkat dari 1.243.103 pada 2011 menjadi 2.049.576 pada 2012, atau tumbuh sekitar 64,9%. Pada 2012, September menjadi bulan dengan total penyewaan tertinggi, yaitu 218.573, sedangkan Januari menjadi yang terendah, yaitu 96.744. Rata-rata penyewaan hariannya sekitar 7.285,77 pada September dan 3.120,77 pada Januari. Perbedaan ini menunjukkan bahwa permintaan berubah sepanjang tahun. Rata-rata harian juga perlu diperhatikan agar perbandingan antarbulan tidak hanya dipengaruhi oleh perbedaan jumlah hari.

**2. Hubungan Kondisi Lingkungan dengan Penyewaan**

Cuaca cerah atau sedikit berawan memiliki rata-rata penyewaan tertinggi, sekitar 4.876,79 per hari. Kondisi sedikit berkabut atau berawan memiliki rata-rata sekitar 4.035,86, sedangkan kondisi hujan atau salju ringan sekitar 1.803,29. Korelasi Pearson antara suhu dan jumlah penyewaan sekitar 0,63, yang menunjukkan hubungan positif pada data ini. Korelasi kelembapan dengan jumlah penyewaan sekitar -0,10, yang menunjukkan hubungan negatif tetapi relatif lemah. Hasil ini menggambarkan asosiasi, bukan bukti bahwa cuaca menjadi satu-satunya penyebab perubahan permintaan.

**3. Pola Penyewaan Berdasarkan Jam dan Jenis Hari**

Pada hari kerja, rata-rata penyewaan tertinggi terjadi pukul 17.00, sekitar 525,29 penyewaan per jam. Rata-rata pada pukul 08.00 juga tinggi, sekitar 477,00, dan pukul 18.00 sekitar 492,20. Pada hari nonkerja, puncak penyewaan terjadi pukul 13.00, sekitar 372,73. Perbedaan pola ini menunjukkan bahwa waktu kebutuhan sepeda pada hari kerja dan hari nonkerja tidak sama. Pengelola perlu menyiapkan armada sebelum periode puncak masing-masing jenis hari.

**4. Pola Penyewaan Berdasarkan Musim**

Musim Fall memiliki rata-rata penyewaan tertinggi, sekitar 5.644,30 per hari, dengan total 1.061.129 penyewaan. Musim Spring memiliki rata-rata terendah, sekitar 2.604,13 per hari, dengan total 471.348 penyewaan. Perbedaan tersebut menunjukkan adanya variasi permintaan musiman yang perlu dipertimbangkan dalam perencanaan distribusi armada dan pemeliharaan.

### 6.2 Recommendation / Action Items

1. **Menyesuaikan distribusi armada secara musiman dan bulanan.** Pengelola dapat menyiapkan kapasitas sepeda lebih besar pada bulan dan musim dengan rata-rata penyewaan tinggi. Evaluasi dilakukan secara berkala menggunakan total dan rata-rata harian agar keputusan tidak hanya bergantung pada volume bulanan.

2. **Menyiapkan sepeda sebelum jam sibuk.** Pada hari kerja, pemeriksaan dan redistribusi armada perlu dilakukan sebelum pukul 08.00 dan 17.00–18.00. Pada hari nonkerja, kesiapan armada perlu ditingkatkan menjelang pukul 13.00.

3. **Memasukkan kondisi cuaca dalam perencanaan operasional.** Informasi cuaca dapat digunakan sebagai salah satu masukan untuk memperkirakan perubahan permintaan. Pengelola sebaiknya tetap mempertimbangkan faktor lain karena hubungan statistik tidak membuktikan sebab-akibat.

4. **Menjadwalkan pemeliharaan pada periode permintaan relatif rendah.** Data menunjukkan rata-rata penyewaan lebih rendah pada musim Spring dibandingkan Fall. Periode dengan permintaan lebih rendah dapat dipertimbangkan untuk pemeliharaan terencana, dengan tetap memperhatikan variasi permintaan bulanan dan operasional setempat.
